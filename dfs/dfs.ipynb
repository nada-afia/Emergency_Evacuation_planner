```python
import time
building = [
    ["S", "A", "B", "X", "C"],
    ["D", "F", "E", "K", "G"],
    ["H", "I", "J", "L", "M"]
]

fires = ["X", "F", "K"]

nodes = {
    "S": ["A", "D"],
    "A": ["S", "B"],
    "B": ["A", "C", "E"],
    "C": ["B", "G"],      
    "D": ["S", "E", "H"], 
    "E": ["B", "D", "J"], 
    "G": ["C", "M"],      
    "H": ["D", "I"],
    "I": ["H", "J"],      
    "J": ["E", "I", "L"],
    "L": ["J", "M"],      
    "M": ["G", "L"],
}

start = "S"
goal = "E"
```

```python
def dfs_fire(nodes, start, goal, fires):
    start_time = time.time()
    stack = [(start, [start])]
    visited = set()
    expanded_order = []

    while stack:
        node, path = stack.pop()
        if node in visited:
            continue
        visited.add(node)
        expanded_order.append(node)
        print(f"Expanded: {node}")

        if node == goal:
            end_time = time.time()
            return path, expanded_order, end_time - start_time

        for neigh in reversed(nodes.get(node, [])):
            if neigh not in visited and neigh not in fires:
                stack.append((neigh, path + [neigh]))

    end_time = time.time()
    return None, expanded_order, end_time - start_time
print("\n===== DFS Result =====")
path, expanded, runtime = dfs_fire(nodes, start, goal, fires)
print("Expanded Nodes:", " -> ".join(expanded))
if path:
    print("Path:", " -> ".join(path))
    print("Path Length:", len(path)-1)
else:
    print("No safe path found")
print(f"Time Taken: {runtime:.6f} seconds")
```
