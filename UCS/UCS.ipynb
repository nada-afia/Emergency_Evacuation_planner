```python
def path_cost(path):
    return sum(step[1] for step in path)
```

```python
def ucs_fire(nodes, start, goal, fires):
    start_time = time.time()
    pq = []
    heapq.heappush(pq, (0, start, [start]))  # (cost, node, path)
    visited = set()
    expanded_order = []

    while pq:
        cost, node, path = heapq.heappop(pq)
        if node in visited:
            continue
        visited.add(node)
        expanded_order.append(node)
        print(f"Expanded: {node} | Cost so far = {cost}")

        if node == goal:
            end_time = time.time()
            return path, cost, expanded_order, end_time - start_time

        for neigh in nodes.get(node, []):
            if neigh not in visited and neigh not in fires:
                heapq.heappush(pq, (cost + 1, neigh, path + [neigh]))

    end_time = time.time()
    return None, None, expanded_order, end_time - start_time


path, cost, expanded, runtime = ucs_fire(nodes, start, goal, fires)

print("\n===== UCS Result =====")
print("Expanded Nodes:", " -> ".join(expanded))
if path:
    print("Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No safe path found")
print(f"Time Taken: {runtime:.6f} seconds")
```
